## 📌 Бизнес-контекст и методология
В 2027 году 85% первичного скрининга резюме автоматизировано AI-агентами (ATS).
Чтобы попасть на интервью к Hiring Manager, резюме должно отвечать трем критериям:
1. **Google XYZ Formula:** *«Достиг [X], измеренного метрикой [Y], применив [Z]»*.
2. **Плотность метрик:** не менее 2 бизнес-метрик ($, %, часов) на каждый пункт опыта.
3. **Proof-of-Work:** рабочие ссылки на воспроизводимые Colab-тетради и GitHub-проекты.

```python
# 🚀 Универсальная инициализация окружения и данных
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
PRIMARY = "#1779ba"
SUCCESS = "#3adb76"
ALERT = "#cc4b37"
WARNING = "#ffae00"
SECONDARY = "#767676"

# Примеры двух резюме: классическое слабое vs оптимизированное по XYZ-стандарту 2027
resume_weak = """
- Писал SQL запросы к базе данных клиентов
- Строил дашборды в Tableau и выгружал отчеты в Excel
- Участвовал в командных созвонах и согласовывал ТЗ
- Проводил базовый анализ данных и строил графики
"""

resume_strong = """
- Сократил отток B2B клиентов на 4.2% ($140k ARR), разработав прогностическую когортную модель в Python и PostgreSQL
- Ускорил Time-to-Insight топ-менеджмента с 3 дней до 15 минут, автоматизировав dbt-витрины данных и Looker Studio дашборд
- Повысил конверсию оформления заказа на +1.8% (+$85k в месяц) через проведение 12 A/B тестов с проверкой CUPED и SRM
- Внедрил агентскую Text-to-SQL систему, сократившую 35 часов рутинных ad-hoc запросов аналитиков в неделю
"""

print("Резюме подготовлены для автоматического NLP-аудита.")
```

## 🔍 Разбор механики: Алгоритм скоринга резюме по формуле Google XYZ
Проанализируем плотность метрик воздействия (числа, проценты, доллары) и глаголов действия.

```python
def score_resume(text):
    lines = [l.strip() for l in text.strip().split('\n') if l.strip().startswith('-')]
    total_bullets = len(lines)
    
    # Регулярные выражения для поиска $ / % / числовых результатов
    metric_pattern = re.compile(r'(\d+[\.,]?\d*\s*(%|\$|k|к|ч|час|дней|минут))', re.IGNORECASE)
    action_verb_pattern = re.compile(r'(сократил|ускорил|повысил|внедрил|оптимизировал|разработал)', re.IGNORECASE)
    
    metrics_count = sum(len(metric_pattern.findall(l)) for l in lines)
    action_bullets = sum(1 for l in lines if action_verb_pattern.search(l))
    
    impact_density = metrics_count / max(total_bullets, 1)
    xyz_compliance = (action_bullets / max(total_bullets, 1)) * 100
    
    overall_score = min(100, int(impact_density * 35 + xyz_compliance * 0.65))
    return {
        'Всего пунктов': total_bullets,
        'Найдено метрик': metrics_count,
        'Плотность метрик (на пункт)': round(impact_density, 2),
        'XYZ соответствие (%)': round(xyz_compliance, 1),
        'Итоговый AI Score (100)': overall_score
    }

audit_df = pd.DataFrame([score_resume(resume_weak), score_resume(resume_strong)], 
                        index=['Слабое резюме (Task-based)', 'Сильное резюме 2027 (XYZ Impact)'])
display(audit_df)
```

## 📊 Визуализация бизнес-кейса: Радар компетенций сильного резюме
Сравним баланс компетенций кандидатов по 5 осям оценки технического директора.

```python
categories = ['Бизнес-метрики ($)', 'SQL & Моделирование', 'A/B & Эксперименты', 'AI Стек & Автоматизация', 'Data Storytelling']
weak_vals = [20, 40, 20, 10, 30]
strong_vals = [95, 90, 85, 90, 85]

N = len(categories)
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]
weak_vals += weak_vals[:1]
strong_vals += strong_vals[:1]

fig, ax = plt.subplots(figsize=(7, 6), subplot_kw=dict(polar=True), dpi=110)
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=9.5, fontweight='bold')
ax.set_ylim(0, 100)

ax.plot(angles, weak_vals, color=SECONDARY, linewidth=1.5, label='Task-based профиль')
ax.fill(angles, weak_vals, color=SECONDARY, alpha=0.15)

ax.plot(angles, strong_vals, color=SUCCESS, linewidth=2.5, label='AI Analytics Engineer 2027')
ax.fill(angles, strong_vals, color=SUCCESS, alpha=0.2)

ax.set_title("Профиль резюме: Влияние доказанных бизнес-результатов", fontsize=11.5, fontweight='bold', pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), frameon=True)

plt.tight_layout()
plt.show()
```
