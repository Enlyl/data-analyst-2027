## 📌 Бизнес-контекст и методология
Среда разработки аналитика эволюционировала от простых блокнотов к специализированным AI IDE (Cursor, Windsurf).
Ключевые сценарии продуктивности:
- **Multi-file Context Editing:** одновременное изменение SQL-модели и связанного Python-скрипта валидации
- **Docs Indexing:** индексация локальной документации репозитория через `@docs`
- **Fast Tab Autocomplete:** предиктивное автодополнение длинных названий колонок и агрегаций

```python
# 🚀 Универсальная инициализация окружения и данных
import os
import urllib.request
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']
plt.rcParams['axes.edgecolor'] = '#e6e6e6'
plt.rcParams['axes.linewidth'] = 0.8

PRIMARY = "#1779ba"
SUCCESS = "#3adb76"
ALERT = "#cc4b37"
WARNING = "#ffae00"
SECONDARY = "#767676"
```

## 🔍 Разбор механики: Сравнение инструментов разработки аналитика
Сравним скорость выполнения комплексных аналитических задач при разных инструментальных сетапах.

```python
# Демонстрация: Скорость разработки аналитического пайплайна
tools_comp = pd.DataFrame({
    'Инструмент': ['Классический VS Code / Jupyter', 'GitHub Copilot в IDE', 'Cursor AI (Agent Mode)'],
    'Создание ETL скрипта (мин)': [45, 20, 8],
    'Отладка и тесты (мин)': [30, 15, 5],
    'Оформление документации (мин)': [25, 10, 2]
})
tools_comp['Общее время (мин)'] = tools_comp.sum(axis=1, numeric_only=True)
display(tools_comp)
```

## 📊 Визуализация бизнес-кейса: Ускорение разработки аналитических пайплайнов
Построим сравнение затрат времени по этапам разработки.

```python
# Визуализация продуктивности AI IDE
fig, ax = plt.subplots(figsize=(9, 4.8), dpi=110)

tools = tools_comp['Инструмент']
total_times = tools_comp['Общее время (мин)']

bars = ax.bar(tools, total_times, color=[SECONDARY, PRIMARY, SUCCESS], edgecolor='#333333', linewidth=0.6, width=0.5)

ax.set_title("Суммарное время создания аналитического пайплайна (минуты)", fontsize=12, fontweight='bold', pad=15)
ax.set_ylabel("Затраченное время (мин)", fontsize=11)
ax.set_ylim(0, 120)

for b in bars:
    h = b.get_height()
    ax.annotate(f"{h} мин", xy=(b.get_x() + b.get_width()/2, h), xytext=(0, 4),
                 textcoords="offset points", ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

### 🎯 Практическое задание 1: Расчет кратности ускорения
Во сколько раз Cursor AI ускоряет разработку по сравнению с классическим подходом?

```python
### ВАШ КОД ЗДЕСЬ
classic_time = tools_comp.loc[0, 'Общее время (мин)']
cursor_time = tools_comp.loc[2, 'Общее время (мин)']
speedup = classic_time / cursor_time

print(f"Кратность ускорения разработки: {speedup:.1f}x!")

assert speedup > 5, "Ускорение должно превышать 5 раз"
print("✅ Тест успешно пройден: эффект внедрения современного AI-стека очевиден!")
```
